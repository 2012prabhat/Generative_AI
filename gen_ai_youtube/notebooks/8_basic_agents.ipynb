```python
from dotenv import load_dotenv
load_dotenv
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
        it will return sum of two numbers
        
        Args:
            a : Number One
            b : Number Two
    """
    return a + b


@tool
def product_numbers(a:int, b:int):
    """
        it will return product of two numbers
        
        Args:
            a : Number One
            b : Number Two
    """
    return a * b
```

```python
add_numbers.invoke({"a":12, "b":3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
agent = create_agent(model=llm,tools=[add_numbers, product_numbers],system_prompt="You are a math teacher, and always use tool for calculation.")
```

```python
response = agent.invoke({'messages':[{"role":"user", "content":"What is 2+3*89?"}]})
print(response["messages"][-1].content)
```
