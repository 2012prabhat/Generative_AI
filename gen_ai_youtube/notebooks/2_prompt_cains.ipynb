# Prompt Templates - Dynamic Prompts


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm = ChatGoogleGenerativeAI(model="gemini-3-flash-preview")
out = StrOutputParser()

prompts = ChatPromptTemplate.from_messages([
    {'role':'system','content':'You are a translator and translate input into {language}'},
    {'role':'user','content':'{query}'}
])

# final_prompt = prompts.format_messages(language="Hinglish", query="I love python and javascript")
# res = llm.invoke(final_prompt)
# print(res.content)

def transform_case(result:str):
    return result.upper()

chains = prompts | llm | out | transform_case
res = chains.invoke({"language":"Hinglish","query":"Langchain is best library"})
print(res)

```
