```python
# Create multi dimensionl arrays using numpy 
np.array([1,2,3,4,5],dtype=np.int32)

arr = [5.1,6.5,7,8.56,9]
np.array(arr,dtype=float)

boolArrr = np.array(arr,dtype=bool)
print(boolArrr)
```

```python
arr3 = np.arange(1,10)
arr3.dtype
arr3 = arr3.reshape(3,3)
max_in_col = np.max(arr3,axis=0)
min_in_col = np.min(arr3,axis=0)

print(f'max value is {max_in_col} and min value is {min_in_col}')

np.sum(arr3,axis=1)
```
