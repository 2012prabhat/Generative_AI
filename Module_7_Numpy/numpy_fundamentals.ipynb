```python
arr2  = np.arange(1,10)
arr2.dtype
arr2 = arr2.reshape(3,3)
arr2.astype('float')
```

```python
arr3 = np.arange(1,10)
arr3.dtype
arr3 = arr3.reshape(3,3)
max_in_col = np.max(arr3,axis=0)
min_in_col = np.min(arr3,axis=0)

print(f'max value is {max_in_col} and min value is {min_in_col}')

np.sum(arr3,axis=1)
np.prod(arr3,axis=1)
```

```python
originalArr = np.arange(1,10).reshape(3,3)
print('Original Arr\n',originalArr)
print('Transpose Arr\n',originalArr.transpose())

np.ravel(originalArr)
```

```python
arr_1 = np.arange(1,13).reshape(3,4)
arr_2 = np.arange(13,25).reshape(3,4)

horizontal_arr = np.hstack((arr_1,arr_2))
print(horizontal_arr)

vertical_arr = np.vstack((arr_1,arr_2))
print(vertical_arr)

# spliting arr 

np.vsplit(vertical_arr,2)

np.hsplit(horizontal_arr,2)
```
