### Task-1 Write Sales records to a file

```python

sales = [1200, 450, 980, 1500, 3000]

try:
   with open("sales_data.txt", "w") as file:
      for index,value in enumerate(sales):
         separator = '\n'
         #optional task 
         # separator = "," if index < len(sales)-1 else ''
         file.write(f'{value}{separator}')
         
   with open('sales_data.txt','r') as file:
      content = file.read()
      print("This is the file content",content)
         
except:
   print('File does not exists')
       
    

```

### Task-3 Append New Sales

```python
new_sales = [5000,2500,1700]
with open('sales_data.txt','a+') as file:
    for sales in new_sales:
        file.write(f'{sales}\n')
    file.seek(0)
    content = file.read()
    print(content)
    # optional - print total number of lines after appending
    print(len(content.split('\n'))-1)
    
```

#### Task-4 Generate Summary Report from file

```python
try:
    with open('sales_data.txt','r') as file:
        all_sales = file.readlines()
        all_sales_numbers = [int(sales.strip()) for sales in all_sales]
        total_sales = sum(all_sales_numbers)
        highest_sales = max(all_sales_numbers)
        lowest_sales = min(all_sales_numbers)
        average_sales = sum(all_sales_numbers) / len(all_sales_numbers)
        print('Total Sales: ',total_sales)
        print('Highest Sales: ',highest_sales)
        print('Lowest Sales: ',lowest_sales)
        print('Average Sales: ',average_sales)
except:
    print('File does not exists')


```

### Task-5 Create Product Info File (User Input)

```python
num_of_products = int(input("Enter the number of products you want to add"))
# products = {}
with open('products.txt','a+') as file:
    file.write('ProductName|Price\n')
    
for i in range(0,num_of_products):
    prod_name = input('Enter product name')
    prod_price = input('Enter product price')
    # products[prod_name] = prod_price
    with open('products.txt','a+') as file:
        file.write(f'{prod_name}|{prod_price}\n')
        
with open('products.txt','r') as file:
    content = [line.strip()  for line in file.readlines()]
    for prod in content:
        prod_name = prod.split('|')[0]
        prod_price = prod.split('|')[1]
        print(prod_name+"   "+prod_price)


    
```

### Task-7 Mini Project - Export Discounted Prices

```python
prices = {
    'Mouse':500,
    'Keyboard': 800,
    'Monitor': 7000,
    'Pendrive': 400,
    'Camera':5000
}

discPer = float(input('Please enter discount percentage.'))
with open("discount_report.txt", "w") as file:
    file.write("Product | Original Price | Discounted Price\n")

total_discPrice = 0
for key,value in prices.items():
    discPrice = value - value * (discPer/100)
    total_discPrice += discPrice
    with open('discount_report.txt','a+') as file:
        file.write(f'{key} | {value} | {discPrice}\n')
        
    
with open('discount_report.txt','a+') as file:
    content = f'\nSummary\nTotal Items - {len(prices)}\nAverage Discounted Price - {total_discPrice/len(prices)}'
    file.write(content)
    
with open('discount_report.txt','r') as file:
    print(file.read())
```
