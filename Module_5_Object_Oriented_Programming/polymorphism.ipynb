```python
class Payment():
    def pay(self):
        pass

class CreditCardPayment(Payment):
   def pay(self):
        print("Paying using Credit Card")

class CashPayment(Payment):
   def pay(self):
        print("Paying using Cash")

class PaypalPayment(Payment):
   def pay(self):
        print("Paying using Paypal")

product1 = PaypalPayment()
product1.pay()

product2 = CashPayment()
product2.pay()

product3 = CreditCardPayment()
product3.pay()
```
