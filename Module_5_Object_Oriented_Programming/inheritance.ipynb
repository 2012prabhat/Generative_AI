```python
class Parent:
    def __init__(self,color,height,nationality):
        self.color = color
        self.height = height
        self.nationality = nationality
        
class Child(Parent):
    def __init__(self,color,height,nationality,occupation):
        super().__init__(color,height,nationality)
        self.occupation = occupation
        
        
child1 = Child('White','Tall','American','Engineer')
print(child1.color)
```
