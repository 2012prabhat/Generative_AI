```python
# instance methods are functions that are defined inside a class and can be called on an instance of that class. They typically operate on the data contained in the instance (i.e., the attributes) and can perform actions or calculations based on that data.
class Avenger:
    origin = 'Marvel'
    def __init__(this,name,power,nationality,weapon):
        this.name = name
        this.power = power
        this.nationality = nationality
        this.weapon = weapon
    
    def attack(this):
       print(f'{this.name} attacks with {this.power} and he is {this.nationality} and he uses {this.weapon}')

    @classmethod
    def get_origin(cls):
        return cls.origin
    
    @staticmethod
    def assemble():
        print('All avengers assemble')


avenger1 = Avenger('Ironman','flying','american','robotic suit')
avenger1.attack()
print(Avenger.get_origin())
Avenger.assemble()
avenger1.assemble()



```
