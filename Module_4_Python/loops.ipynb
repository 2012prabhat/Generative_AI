```python
tupList = (1,2,3,4,5,6,7)
print(tupList[::-1])
print(tupList)

print(tupList[2])

print(tupList.index(3))

test = 1,2,3,4,
print(type(test))
print(type(test)==tuple)
```

```python
# un packing of tuples
a,b,c,d,e = 1,2,3,4,5
print(a,b,c,d,e)
```

# Unpacking with *

```python
a,*b,c = 1,2,3,4,5,6,7
print(a)
print(b)
print(c)
```

```python
nestedList = [[1,2,3,4],(5,6,7,5),[10,12,43],(8,8,7,5,4),[1,2,3,4]]
for listElem in nestedList:
    if type(listElem) == tuple:
        print(listElem)
```
